```python
bureau_work = df_bureau.copy()
```

```python

bureau_work["DEBT_NEGATIVE_FLAG"] = (
    bureau_work["AMT_CREDIT_SUM_DEBT"] < 0
).astype(int)


bureau_work["DEBT_NEGATIVE_ABS"] = (
    bureau_work["AMT_CREDIT_SUM_DEBT"]
    .where(bureau_work["AMT_CREDIT_SUM_DEBT"] < 0)
    .abs()
)


bureau_work["DEBT_NONNEGATIVE"] = (
    bureau_work["AMT_CREDIT_SUM_DEBT"]
    .where(bureau_work["AMT_CREDIT_SUM_DEBT"] >= 0)
)
```

```python
currency_summary = (
    bureau_work.groupby("CREDIT_CURRENCY", dropna=False)
    .agg(
        TOTAL_RECORDS=("SK_ID_BUREAU", "count"),
        UNIQUE_CUSTOMERS=("SK_ID_CURR", "nunique"),
        NEGATIVE_DEBT_RECORDS=("DEBT_NEGATIVE_FLAG", "sum")
    )
)

currency_summary["PERCENTAGE"] = (
    currency_summary["TOTAL_RECORDS"] /
    currency_summary["TOTAL_RECORDS"].sum() * 100
)

display(currency_summary.round(2))
```

```python
bureau_features = (
    bureau_work.groupby("SK_ID_CURR")
    .agg(
        # Credit exposure
        BUREAU_TOTAL_CREDITS=("SK_ID_BUREAU", "nunique"),

        BUREAU_ACTIVE_CREDITS=(
            "CREDIT_ACTIVE",
            lambda x: (x == "Active").sum()
        ),

        BUREAU_CLOSED_CREDITS=(
            "CREDIT_ACTIVE",
            lambda x: (x == "Closed").sum()
        ),

        BUREAU_SOLD_CREDITS=(
            "CREDIT_ACTIVE",
            lambda x: (x == "Sold").sum()
        ),

        BUREAU_BAD_DEBT_CREDITS=(
            "CREDIT_ACTIVE",
            lambda x: (x == "Bad debt").sum()
        ),

        # Payment behavior
        BUREAU_OVERDUE_CREDIT_COUNT=(
            "CREDIT_DAY_OVERDUE",
            lambda x: (x > 0).sum()
        ),

        BUREAU_MAX_OVERDUE_DAYS=(
            "CREDIT_DAY_OVERDUE",
            "max"
        ),

        BUREAU_TOTAL_PROLONG=(
            "CNT_CREDIT_PROLONG",
            "sum"
        ),

        # Data quality
        BUREAU_NEGATIVE_DEBT_RECORDS=(
            "DEBT_NEGATIVE_FLAG",
            "sum"
        )
    )
)
```

```python
monetary_columns = {
    "AMT_CREDIT_SUM": "CREDIT_SUM",
    "DEBT_NONNEGATIVE": "DEBT",
    "DEBT_NEGATIVE_ABS": "NEGATIVE_DEBT_ABS",
    "AMT_CREDIT_SUM_OVERDUE": "OVERDUE_AMOUNT"
}

currencies = sorted(
    bureau_work["CREDIT_CURRENCY"].dropna().unique()
)

for currency in currencies:

    currency_data = bureau_work[
        bureau_work["CREDIT_CURRENCY"] == currency
    ]

    currency_name = currency.upper().replace(" ", "_")

    # Number of credits per currency
    currency_count = (
        currency_data.groupby("SK_ID_CURR")["SK_ID_BUREAU"]
        .nunique()
    )

    bureau_features = bureau_features.join(
        currency_count.rename(
            f"BUREAU_{currency_name}_CREDIT_COUNT"
        )
    )

    # Financial aggregation per currency
    for source_col, feature_name in monetary_columns.items():

        aggregated = (
            currency_data.groupby("SK_ID_CURR")[source_col]
            .sum(min_count=1)
        )

        bureau_features = bureau_features.join(
            aggregated.rename(
                f"BUREAU_{currency_name}_{feature_name}"
            )
        )
```

```python
bureau_analysis = (
    df_train[["SK_ID_CURR", "TARGET"]]
    .merge(
        bureau_features.reset_index(),
        on="SK_ID_CURR",
        how="left",
        validate="one_to_one"
    )
)

bureau_analysis["HAS_BUREAU_HISTORY"] = (
    bureau_analysis["BUREAU_TOTAL_CREDITS"]
    .notna()
    .astype(int)
)

print("Analysis Dataset:", bureau_analysis.shape)

print(
    "Customers with Bureau History:",
    bureau_analysis["HAS_BUREAU_HISTORY"].sum()
)

print(
    "Customers without Bureau History:",
    (bureau_analysis["HAS_BUREAU_HISTORY"] == 0).sum()
)
```

```python
comparison_features = [
    "BUREAU_TOTAL_CREDITS",
    "BUREAU_ACTIVE_CREDITS",
    "BUREAU_CLOSED_CREDITS",
    "BUREAU_OVERDUE_CREDIT_COUNT",
    "BUREAU_MAX_OVERDUE_DAYS",
    "BUREAU_TOTAL_PROLONG",
    "BUREAU_NEGATIVE_DEBT_RECORDS"
]

bureau_target_comparison = (
    bureau_analysis.groupby("TARGET")[comparison_features]
    .median()
    .T
)

bureau_target_comparison.columns = [
    "No Payment Difficulties",
    "Payment Difficulties"
]

display(bureau_target_comparison.round(2))
```

```python
coverage_analysis = (
    bureau_analysis.groupby("TARGET")
    .agg(
        TOTAL_APPLICATIONS=("SK_ID_CURR", "count"),
        WITH_BUREAU=("HAS_BUREAU_HISTORY", "sum")
    )
)

coverage_analysis["WITHOUT_BUREAU"] = (
    coverage_analysis["TOTAL_APPLICATIONS"]
    - coverage_analysis["WITH_BUREAU"]
)

coverage_analysis["BUREAU_COVERAGE_RATE"] = (
    coverage_analysis["WITH_BUREAU"]
    / coverage_analysis["TOTAL_APPLICATIONS"] * 100
)

display(coverage_analysis.round(2))
```

```python
bureau_with_history = bureau_analysis[
    bureau_analysis["HAS_BUREAU_HISTORY"] == 1
].copy()

bureau_with_history["HAS_CURRENT_OVERDUE"] = (
    bureau_with_history["BUREAU_OVERDUE_CREDIT_COUNT"] > 0
).astype(int)

bureau_with_history["HAS_ACTIVE_CREDIT"] = (
    bureau_with_history["BUREAU_ACTIVE_CREDITS"] > 0
).astype(int)

bureau_with_history["HAS_CLOSED_CREDIT"] = (
    bureau_with_history["BUREAU_CLOSED_CREDITS"] > 0
).astype(int)

bureau_with_history["HAS_NEGATIVE_DEBT"] = (
    bureau_with_history["BUREAU_NEGATIVE_DEBT_RECORDS"] > 0
).astype(int)

risk_comparison = (
    bureau_with_history.groupby("TARGET")
    .agg(
        CUSTOMERS=("SK_ID_CURR", "count"),

        AVG_TOTAL_CREDITS=("BUREAU_TOTAL_CREDITS", "mean"),
        AVG_ACTIVE_CREDITS=("BUREAU_ACTIVE_CREDITS", "mean"),
        AVG_CLOSED_CREDITS=("BUREAU_CLOSED_CREDITS", "mean"),

        PCT_HAS_ACTIVE_CREDIT=("HAS_ACTIVE_CREDIT", "mean"),
        PCT_HAS_CLOSED_CREDIT=("HAS_CLOSED_CREDIT", "mean"),
        PCT_HAS_CURRENT_OVERDUE=("HAS_CURRENT_OVERDUE", "mean"),

        AVG_MAX_OVERDUE_DAYS=("BUREAU_MAX_OVERDUE_DAYS", "mean"),

        MEDIAN_CREDIT_SUM=("BUREAU_CURRENCY_1_CREDIT_SUM", "median"),
        MEDIAN_DEBT=("BUREAU_CURRENCY_1_DEBT", "median")
    )
)

percentage_columns = [
    "PCT_HAS_ACTIVE_CREDIT",
    "PCT_HAS_CLOSED_CREDIT",
    "PCT_HAS_CURRENT_OVERDUE"
]

risk_comparison[percentage_columns] *= 100

display(risk_comparison.round(2))
```

```python
status_counts = (
    matched_balance
    .groupby(
        ["SK_ID_BUREAU", "STATUS"],
        observed=True
    )
    .size()
    .unstack(fill_value=0)
)

status_counts.columns = status_counts.columns.astype(str)

# Ensure all expected status categories exist
status_order = ["0", "1", "2", "3", "4", "5", "C", "X"]

status_counts = status_counts.reindex(
    columns=status_order,
    fill_value=0
)

display(status_counts.head())
```

```python
bb_credit_features = pd.DataFrame(
    index=status_counts.index
)

# Total monthly observations
bb_credit_features["BB_TOTAL_MONTHS"] = (
    status_counts.sum(axis=1)
)

# Months with informative credit status (0-5)
valid_status = ["0", "1", "2", "3", "4", "5"]

bb_credit_features["BB_VALID_MONTHS"] = (
    status_counts[valid_status].sum(axis=1)
)

# Any overdue status (1-5)
dpd_status = ["1", "2", "3", "4", "5"]

bb_credit_features["BB_DPD_MONTHS"] = (
    status_counts[dpd_status].sum(axis=1)
)

# Severe overdue status (3-5)
severe_status = ["3", "4", "5"]

bb_credit_features["BB_SEVERE_DPD_MONTHS"] = (
    status_counts[severe_status].sum(axis=1)
)

# Closed and unknown status
bb_credit_features["BB_CLOSED_MONTHS"] = (
    status_counts["C"]
)

bb_credit_features["BB_UNKNOWN_MONTHS"] = (
    status_counts["X"]
)

# Maximum observed numeric overdue status (0-5)
numeric_status = status_counts[valid_status]

status_present = numeric_status.gt(0)

max_status = (
    status_present.iloc[:, ::-1]
    .idxmax(axis=1)
    .astype(float)
)

bb_credit_features["BB_MAX_DPD_STATUS"] = (
    max_status.where(status_present.any(axis=1))
)

# Overdue ratio
bb_credit_features["BB_DPD_RATIO"] = (
    bb_credit_features["BB_DPD_MONTHS"] /
    bb_credit_features["BB_VALID_MONTHS"].replace(0, np.nan)
)

display(bb_credit_features.head())
```

```python
# Map credit-level features to customer IDs
bureau_balance_customer = (
    df_bureau[["SK_ID_CURR", "SK_ID_BUREAU"]]
    .merge(
        bb_credit_features.reset_index(),
        on="SK_ID_BUREAU",
        how="left",
        validate="one_to_one"
    )
)

# Identify credits with monthly history
bureau_balance_customer["BB_HAS_HISTORY"] = (
    bureau_balance_customer["BB_TOTAL_MONTHS"]
    .notna()
    .astype(int)
)

bureau_balance_customer["BB_HAS_DPD"] = (
    bureau_balance_customer["BB_DPD_MONTHS"]
    .gt(0)
    .fillna(False)
    .astype(int)
)
```

```python
bureau_balance_features = (
    bureau_balance_customer
    .groupby("SK_ID_CURR")
    .agg(
        BUREAU_BB_CREDITS_WITH_HISTORY=("BB_HAS_HISTORY", "sum"),

        BUREAU_BB_CREDITS_WITH_DPD=("BB_HAS_DPD", "sum"),

        BUREAU_BB_TOTAL_MONTHS=("BB_TOTAL_MONTHS", "sum"),

        BUREAU_BB_VALID_MONTHS=("BB_VALID_MONTHS", "sum"),

        BUREAU_BB_DPD_MONTHS=("BB_DPD_MONTHS", "sum"),

        BUREAU_BB_SEVERE_DPD_MONTHS=(
            "BB_SEVERE_DPD_MONTHS", "sum"
        ),

        BUREAU_BB_CLOSED_MONTHS=("BB_CLOSED_MONTHS", "sum"),

        BUREAU_BB_UNKNOWN_MONTHS=("BB_UNKNOWN_MONTHS", "sum"),

        BUREAU_BB_MAX_DPD_STATUS=("BB_MAX_DPD_STATUS", "max")
    )
)
```

```python
bureau_balance_features["BUREAU_BB_DPD_RATIO"] = (
    bureau_balance_features["BUREAU_BB_DPD_MONTHS"] /
    bureau_balance_features["BUREAU_BB_VALID_MONTHS"].replace(0, np.nan)
)

bureau_balance_features["BUREAU_BB_SEVERE_DPD_RATIO"] = (
    bureau_balance_features["BUREAU_BB_SEVERE_DPD_MONTHS"] /
    bureau_balance_features["BUREAU_BB_VALID_MONTHS"].replace(0, np.nan)
)

bureau_balance_features["BUREAU_BB_CREDIT_DPD_RATIO"] = (
    bureau_balance_features["BUREAU_BB_CREDITS_WITH_DPD"] /
    bureau_balance_features["BUREAU_BB_CREDITS_WITH_HISTORY"].replace(0, np.nan)
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Identifikasi tipe fitur
numeric_features = X_train_model.select_dtypes(
    include=['number']
).columns.tolist()

categorical_features = X_train_model.select_dtypes(
    include=['object', 'category', 'bool', 'str']
).columns.tolist()

print("Fitur numerik:", len(numeric_features))
print("Fitur kategorikal:", len(categorical_features))

# Pipeline numerik
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(
        strategy='median',
        add_indicator=True
    )),
    ('scaler', StandardScaler())
])

# Pipeline kategorikal
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Gabungkan preprocessing
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from lightgbm import LGBMClassifier
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)

# Identifikasi tipe fitur berdasarkan training set
numeric_features_final = X_train_final.select_dtypes(
    include=['number']
).columns.tolist()

categorical_features_final = X_train_final.select_dtypes(
    include=['object', 'category', 'bool', 'str']
).columns.tolist()

print("Fitur numerik:", len(numeric_features_final))
print("Fitur kategorikal:", len(categorical_features_final))

# Preprocessing numerik
numeric_transformer_final = Pipeline(steps=[
    ('imputer', SimpleImputer(
        strategy='median',
        add_indicator=True
    )),
    ('scaler', StandardScaler())
])

# Preprocessing kategorikal
categorical_transformer_final = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Gabungkan preprocessing
preprocessor_final = ColumnTransformer(transformers=[
    ('num', numeric_transformer_final, numeric_features_final),
    ('cat', categorical_transformer_final, categorical_features_final)
])

# Hitung bobot kelas dari training set saja
scale_pos_weight_final = (
    (y_train_final == 0).sum()
    / (y_train_final == 1).sum()
)

# Buat pipeline LightGBM baru
model_lgbm_final = Pipeline(steps=[
    ('preprocessor', preprocessor_final),
    ('classifier', LGBMClassifier(
        objective='binary',
        n_estimators=500,
        learning_rate=0.05,
        num_leaves=31,
        max_depth=-1,
        min_child_samples=50,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        scale_pos_weight=scale_pos_weight_final,
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    ))
])

# Training ulang
model_lgbm_final.fit(X_train_final, y_train_final)

print("Training LightGBM final selesai.")
```

```python
def save_fairness_plot(data, metric, title, filename):

    plot_data = data[data['SUPPORT'] >= 30].copy()

    pivot_data = plot_data.pivot(
        index='GROUP',
        columns='MODEL',
        values=metric
    )

    fig, ax = plt.subplots(figsize=(9, 5))

    pivot_data.plot(
        kind='bar',
        ax=ax
    )

    ax.set_title(title)
    ax.set_ylabel(metric.replace('_', ' ').title())
    ax.set_xlabel('Customer Group')
    ax.set_ylim(0, 1)

    plt.xticks(rotation=0)
    plt.legend(title='Model')
    plt.tight_layout()

    plt.savefig(
        IMAGE_DIR / filename,
        dpi=200,
        bbox_inches='tight'
    )

    plt.show()
    plt.close()


# Gender
save_fairness_plot(
    gender_comparison,
    'RECALL',
    'Recall Comparison by Gender',
    'fairness_gender_recall.png'
)

save_fairness_plot(
    gender_comparison,
    'FALSE_POSITIVE_RATE',
    'False Positive Rate Comparison by Gender',
    'fairness_gender_fpr.png'
)

# Age
save_fairness_plot(
    age_comparison,
    'RECALL',
    'Recall Comparison by Age Group',
    'fairness_age_recall.png'
)

save_fairness_plot(
    age_comparison,
    'FALSE_POSITIVE_RATE',
    'False Positive Rate Comparison by Age Group',
    'fairness_age_fpr.png'
)
```
